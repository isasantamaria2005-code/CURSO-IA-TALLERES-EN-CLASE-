##Revisión de duplicados
 Se revisó la existencia de registros duplicados en el conjunto de datos mediante la función duplicated(). Los registros completamente duplicados se identifican y posteriormente se eliminan con drop_duplicates(), con el fin de evitar que una misma observación sea contabilizada más de una vez. Sin embargo, que se repita el id_vivienda no significa necesariamente que exista un duplicado. Este identificador corresponde a la vivienda, por lo que puede aparecer varias veces cuando en una misma vivienda hay diferentes personas u hogares registrados. Por esta razón, los registros con id_vivienda repetido no deben eliminarse automáticamente.

# Ejercicio 9: Groupby

Agrupar los datos por categorías, nos permite realizar análisis más profundos y obtener insights valiosos, que a simple vista no vemos en la base de datos.

1. Crea un nuevo dataframe llamado "df_sexo" donde se agrupe por 'sexo' y se calcule la suma de las columnas 'poblacion_total' y 'ocupados'. Crea una nueva columna que se llame 'porcentaje_ocupados' donde dividas las columna 'ocupados' sobre 'poblacion_total'. Al final usa reset_index() para convertir las categorías de sexo en una columna.

```python
df_sexo = (
    df_empleo
    .groupby('sexo', observed=True)[['poblacion_total', 'ocupados']]
    .sum()
    .reset_index()
)

df_sexo['porcentaje_ocupados'] = (
    df_sexo['ocupados'] / df_sexo['poblacion_total']
) * 100

df_sexo
```

2. Crea un nuevo dataframe llamado "df_educacion" donde se agrupe por 'nivel_educativo' y se calcule la suma de las columnas 'poblacion_total' y 'ocupados'. Crea una nueva columna que se llame 'porcentaje_ocupados' donde dividas las columna 'ocupados' sobre 'poblacion_total'. Al final usa reset_index() para convertir las categorías de nivel educativo en una columna.

```python
df_educacion = (
    df_empleo
    .groupby('nivel_educativo', observed=True)[['poblacion_total', 'ocupados']]
    .sum()
    .reset_index()
)

df_educacion['porcentaje_ocupados'] = (
    df_educacion['ocupados'] / df_educacion['poblacion_total']
) * 100

df_educacion
```

# Ejercicio 10: Filtrar datos

1. A partir de df_empleo, crea el DataFrame `df_ing` con observaciones donde `ingreso_laboral` no sea nulo y además sea mayor estrictamente al salario mínimo para 2026 (con subsidio de transporte).

2. ¿Cuántas personas tienen un salario estrictamente mayor al salario mínimo?

```python
niveles_media_superior = [
    'Educación media',
    'Técnica profesional',
    'Tecnológica',
    'Universitaria',
    'Especialización',
    'Maestría',
    'Doctorado'
]
```

3. Crea el dataframe df_mujeres donde se encuentren solamente mujeres que están ubicadas en el departamento del Tolima, que son asalariadas y cuyo nivel educativo es educación media o superior.

```python
df_mujeres = df_empleo[
    (df_empleo['sexo'] == 'Mujer') &
    (df_empleo['nombre_dpto'] == 'Tolima') &
    (df_empleo['categoria_ocupacional'] == 'Asalariado') &
    (df_empleo['nivel_educativo'].isin(niveles_media_superior))
].copy()

df_mujeres.shape
```
